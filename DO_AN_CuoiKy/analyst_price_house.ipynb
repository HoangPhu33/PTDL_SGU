```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Thiết lập trực quan
sns.set(style="whitegrid")
plt.rcParams['figure.dpi'] = 120

# --- Tiền xử lý nhẹ ---
# Chuyển Price về float (tỷ) nếu cần
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

# Nếu muốn loại bỏ những giá trị quá bất thường (optional)
# Ví dụ loại bỏ top 0.5% giá quá cao để biểu đồ dễ nhìn
price_q_low, price_q_high = df['Price'].quantile([0.005, 0.995])
df_viz = df[(df['Price'] >= price_q_low) & (df['Price'] <= price_q_high)].copy()

# Điền tạm NaN cho cột số (nếu có) để dễ vẽ
num_cols = ['Area','Bedrooms','Bathrooms','Floors','Frontage','Access Road']
for c in num_cols:
    if c in df_viz.columns:
        df_viz[c] = pd.to_numeric(df_viz[c], errors='coerce')


# ============================
# 2) Phân tích đặc trưng vật lý
# ============================
# 2.1 Area vs Price (scatter + trendline + hexbin để thấy mật độ)
plt.figure(figsize=(8,5))
sns.regplot(data=df_viz, x='Area', y='Price', scatter_kws={'alpha':0.25, 's':10}, lowess=True, line_kws={'color':'red'})
plt.title("Area vs Price (có đường xu hướng - LOWESS)")
plt.xlabel("Area (m²)")
plt.ylabel("Price (tỷ)")
plt.tight_layout()
plt.show()

# # 2.2 Area bins + mean price (bar + line) — giống bạn đã làm nhưng mở rộng
# bins = [0,40,70,150,300, df_viz['Area'].max()]
# labels = ["0-40","40-70","70-150","150-300","300+"]
# df_viz['Area_bin'] = pd.cut(df_viz['Area'], bins=bins, labels=labels, include_lowest=True)

# area_group = df_viz.groupby('Area_bin')['Price'].agg(['mean','median','count']).reset_index()
# plt.figure(figsize=(9,5))
# sns.barplot(data=area_group, x='Area_bin', y='mean')
# sns.lineplot(data=area_group, x='Area_bin', y='mean', marker='o', color='red')
# plt.title("Giá trung bình theo nhóm diện tích")
# plt.xlabel("Nhóm diện tích (m²)")
# plt.ylabel("Giá trung bình (tỷ)")
# plt.tight_layout()
# plt.show()

# 2.3 Bedrooms / Bathrooms / Floors ảnh hưởng thế nào: dùng boxplot & pointplot cho mean
plt.figure(figsize=(14,4))
plt.subplot(1,3,1)
sns.boxplot(data=df_viz, x='Bedrooms', y='Price')
plt.title("Bedrooms vs Price")
plt.subplot(1,3,2)
sns.boxplot(data=df_viz, x='Bathrooms', y='Price')
plt.title("Bathrooms vs Price")
plt.subplot(1,3,3)
sns.boxplot(data=df_viz, x='Floors', y='Price')
plt.title("Floors vs Price")
plt.tight_layout()
plt.show()

# Ngoài ra vẽ mean+count
bed_stats = df_viz.groupby('Bedrooms')['Price'].agg(['mean','count']).reset_index()
plt.figure(figsize=(7,4))
sns.pointplot(data=bed_stats, x='Bedrooms', y='mean')
plt.title("Giá trung bình theo số Bedrooms")
plt.ylabel("Giá trung bình (tỷ)")
plt.tight_layout()
plt.show()

# 2.4 Frontage & Access Road (scatter + bins)
# Nếu Frontage có nhiều outlier, cân nhắc log hoặc giới hạn
plt.figure(figsize=(8,5))
sns.regplot(data=df_viz, x='Frontage', y='Price', scatter_kws={'alpha':0.25, 's':10}, lowess=True, line_kws={'color':'red'})
plt.title("Frontage vs Price (LOWESS)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8,5))
sns.regplot(data=df_viz, x='Access Road', y='Price', scatter_kws={'alpha':0.25, 's':10}, lowess=True, line_kws={'color':'red'})
plt.title("Access Road vs Price (LOWESS)")
plt.tight_layout()
plt.show()

# Binning Frontage để xem mean price theo nhóm mặt tiền
front_bins = [0,2,4,6,8,15, df_viz['Frontage'].max()]
front_labels = ['<=2','2-4','4-6','6-8','8-15','15+']
df_viz['Frontage_bin'] = pd.cut(df_viz['Frontage'], bins=front_bins, labels=front_labels, include_lowest=True)
front_stats = df_viz.groupby('Frontage_bin')['Price'].agg(['mean','count']).reset_index().dropna()
plt.figure(figsize=(9,4))
sns.barplot(data=front_stats, x='Frontage_bin', y='mean')
plt.title("Giá trung bình theo nhóm Frontage")
plt.xlabel("Frontage (m)")
plt.ylabel("Giá trung bình (tỷ)")
plt.tight_layout()
plt.show()

# ============================
# 3) Phân tích pháp lý & nội thất (categorical)
# ============================
# 3.1 Legal status (mean price + count)
if 'Legal status' in df_viz.columns:
    legal_stats = df_viz.groupby('Legal status')['Price'].agg(['mean','count']).reset_index().sort_values('mean', ascending=False)
    plt.figure(figsize=(9,5))
    sns.barplot(data=legal_stats, x='mean', y='Legal status')
    plt.title("Giá trung bình theo Pháp lý")
    plt.xlabel("Giá trung bình (tỷ)")
    plt.tight_layout()
    plt.show()
    print(legal_stats)

# 3.2 Furniture state (mean price + count)
if 'Furniture state' in df_viz.columns:
    furn_stats = df_viz.groupby('Furniture state')['Price'].agg(['mean','count']).reset_index().sort_values('mean', ascending=False)
    plt.figure(figsize=(9,5))
    sns.barplot(data=furn_stats, x='mean', y='Furniture state')
    plt.title("Giá trung bình theo Tình trạng nội thất")
    plt.xlabel("Giá trung bình (tỷ)")
    plt.tight_layout()
    plt.show()
    print(furn_stats)

# ============================
# 4) Phân phối các cột số — kiểm tra skewness & kurtosis
# ============================
from scipy.stats import skew, kurtosis

num_summary = []
for c in ['Area','Frontage','Floors','Bedrooms','Bathrooms','Price']:
    if c in df_viz.columns:
        ser = df_viz[c].dropna()
        num_summary.append({
            'feature': c,
            'mean': ser.mean(),
            'median': ser.median(),
            'std': ser.std(),
            'skewness': skew(ser),
            'kurtosis': kurtosis(ser)
        })
num_summary = pd.DataFrame(num_summary)
print("\nNumeric summary (skewness/kurtosis):")
print(num_summary)

# Vẽ histogram + kde cho Area & Frontage (bạn nói 2 cột lệch trái)
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
sns.histplot(df_viz['Area'].dropna(), bins=40, kde=True)
plt.title('Area distribution')
plt.subplot(1,2,2)
sns.histplot(df_viz['Frontage'].dropna(), bins=30, kde=True)
plt.title('Frontage distribution')
plt.tight_layout()
plt.show()

# ============================
# 5) Gợi ý xử lý khi biến lệch (advice printed)
# ============================
print("""
Gợi ý (nếu cần tiền xử lý trước modeling):
- Nếu biến số (Area, Frontage) lệch phải mạnh: thử log-transform (np.log1p) hoặc Box-Cox để giảm skew.
- Với biến categorical ít category thì OneHotEncoder ok; nhiều category (ví dụ District nhiều) => dùng target encoding hoặc chỉ lấy top-N và gộp còn lại là 'Other'.
- Với Frontage/Access Road: có thể giữ nguyên (vì có tác động vật lý), nhưng thử thêm biến nhị phân (has_frontage_large = frontage > 6) để capture non-linear effect.
- Với missing nhiều: cân nhắc thêm indicator flag (is_missing_frontage).
""")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# ================================
# 1A. CITY — Mean Price + Count
# ================================
city_stats = (
    df.groupby('City')
      .agg(mean_price=('Price', 'mean'),
           count=('Price', 'count'))
      .sort_values('count', ascending=False)
      .head(20)
      .reset_index()
)

plt.figure(figsize=(12,6))
ax = sns.barplot(
    data=city_stats,  
    x='City',
    y='mean_price',
    palette='viridis'
)

# Ghi số lượng lên cột
for i, row in city_stats.reset_index().iterrows():
    ax.text(
        i,
        row['mean_price'] + 0.05,
        f"{row['count']}",
        ha='center',
        va='bottom',
        fontsize=9
    )

plt.title("Giá trung bình theo City (kèm số lượng BĐS)")
plt.xlabel("City")
plt.ylabel("Giá trung bình (tỷ)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ================================
# 1B. DISTRICT — Mean Price + Count (TOP 20)
# ================================
district_stats = (
    df.groupby('District')
      .agg(mean_price=('Price', 'mean'),
           count=('Price', 'count'))
      .reset_index()
      .sort_values('count', ascending=False)  # sort trước rồi mới lấy top 20
      .head(20)
      .sort_values('count', ascending=False)  # sắp theo giá để vẽ
)

plt.figure(figsize=(14,6))
ax = sns.barplot(
    data=district_stats,
    x='District',
    y='mean_price',
    palette='coolwarm'
)

# Ghi số lượng lên đầu cột
for i, row in district_stats.reset_index().iterrows():
    ax.text(
        i,
        row['mean_price'] + 0.05,
        f"{row['count']}",
        ha='center',
        va='bottom',
        fontsize=8
    )

plt.title("Giá trung bình theo TOP 20 District (kèm số lượng BĐS)")
plt.xlabel("District")
plt.ylabel("Giá trung bình (tỷ)")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------
# 1. Chuẩn bị dữ liệu
# -----------------------------
num_features = ['Area', 'Frontage', 'Access Road', 'Floors', 'Bedrooms', 'Bathrooms', 'Price']

# SỬA Ở ĐÂY: Xóa 2 cột đã bị drop từ trước
cat_features = ['City', 'District', 'Legal status', 'Furniture state']

# Bây giờ sẽ không còn lỗi
df_cluster = df[num_features + cat_features].dropna()

print(f"Số mẫu sau khi dropna: {len(df_cluster)}")
print(f"Số cột categorical sau OneHot: sẽ tăng lên nhiều do có nhiều City/District")

# -----------------------------
# 2. Chuyển categorical sang numeric (OneHot)
# -----------------------------
preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), num_features),
    ('cat', OneHotEncoder(sparse_output=False, handle_unknown='ignore'), cat_features)
])


X_scaled = preprocessor.fit_transform(df_cluster)

# -----------------------------
# 3. Tìm số cụm tối ưu (Elbow + Silhouette)
# -----------------------------
inertia = []
sil_scores = []
# K = range(2, 11)
K = range(4, 30)

for k in K:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X_scaled)
    inertia.append(km.inertia_)
    sil_scores.append(silhouette_score(X_scaled, km.labels_))

# Vẽ kết quả
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].plot(K, inertia, marker='o')
ax[0].set_title("Elbow method")
ax[0].set_xlabel("k")
ax[0].set_ylabel("Inertia")
ax[1].plot(K, sil_scores, marker='o')
ax[1].set_title("Silhouette Score")
ax[1].set_xlabel("k")
ax[1].set_ylabel("Score")
plt.show()

best_k = K[np.argmax(sil_scores)]
print("Best k =", best_k)

# -----------------------------
# 4. Train KMeans cuối
# -----------------------------
kmeans = KMeans(n_clusters=best_k, random_state=42)
df_cluster['cluster'] = kmeans.fit_predict(X_scaled)

# -----------------------------
# 5. Giảm chiều PCA để vẽ 2D
# -----------------------------
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
df_cluster['pca1'] = X_pca[:,0]
df_cluster['pca2'] = X_pca[:,1]

# -----------------------------
# 6. Vẽ scatter plot PCA 2D
# -----------------------------
plt.figure(figsize=(8,6))
sns.scatterplot(data=df_cluster, x='pca1', y='pca2', hue='cluster', palette='Set2', s=50)
plt.title("KMeans Clustering (PCA 2D) với categorical features")
plt.show()

# -----------------------------
# 7. Mô tả cụm numeric
# -----------------------------
cluster_summary_num = df_cluster.groupby('cluster')[num_features].mean()
print(cluster_summary_num)

# ==========================
# 8. Mô tả cụm categorical (sort tỉ lệ giảm dần)
# ==========================
for col in cat_features:
    print(f"\nDistribution of {col} by cluster (sorted):")
    ctab = pd.crosstab(df_cluster['cluster'], df_cluster[col], normalize='index')
    # Sắp xếp các cột theo tỉ lệ giảm dần trong mỗi cluster
    ctab_sorted = ctab.apply(lambda x: x.sort_values(ascending=False).values, axis=1)
    # Hiển thị
    for i, cluster_values in enumerate(ctab_sorted.values):
        print(f"Cluster {i}: {dict(zip(ctab.columns[ctab.iloc[i].argsort()[::-1]], cluster_values))}")

```

```python
from sklearn.tree import DecisionTreeClassifier, export_text
import pandas as pd

# Chọn features numeric + categorical
X_rules = df_cluster[num_features + cat_features].copy()
y_rules = df_cluster['cluster']

# Encode categorical bằng get_dummies
X_encoded = pd.get_dummies(X_rules, columns=cat_features, drop_first=True)

# Train Decision Tree nhỏ
tree = DecisionTreeClassifier(max_depth=3, random_state=42)
tree.fit(X_encoded, y_rules)

# Xuất luật
rules = export_text(tree, feature_names=list(X_encoded.columns))
print(rules)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Chọn các feature numeric để phân cụm
cluster_features = ['Area', 'Frontage', 'Access Road', 'Floors', 'Bedrooms', 'Bathrooms', 'Price']

# Drop NA tạm thời để train KMeans
df_for_cluster = df[cluster_features].dropna()

# Chuẩn hóa
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_for_cluster)

# Train KMeans
best_k = 2
kmeans = KMeans(n_clusters=best_k, random_state=42)
labels = kmeans.fit_predict(X_scaled)

# Tạo cột 'class' trong df gốc
df.loc[df_for_cluster.index, 'class'] = labels

# Kiểm tra kết quả
print(df[['Area', 'Price', 'class']].head(10))

df.to_csv('VN_House_price_Clean2.csv', index=False)
```
