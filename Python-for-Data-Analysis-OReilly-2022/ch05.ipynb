```python
import yfinance as yf
import pandas as pd

tickers = ['AAPL', 'GOOG', 'IBM', 'MSFT']

# Tải dữ liệu một lần, dùng auto_adjust=True (mặc định) để Close đã điều chỉnh
data = yf.download(tickers, start='2010-01-01', end='2016-12-31', progress=False)

# Dùng 'Close' thay vì 'Adj Close' – vì nó đã là giá điều chỉnh!
price = data['Close']      # Đây chính là yahoo_price.pkl trong sách
volume = data['Volume']    # Tương tự yahoo_volume.pkl

# Xem thử
print("Price (Adjusted Close):")
print(price.head())

print("\nVolume:")
print(volume.head())

# Tính returns như trong sách
returns = price.pct_change()
print("\nReturns tail:")
print(returns.tail())

# Các ví dụ tiếp theo trong sách
print("\nCorrelation MSFT vs IBM:")
print(returns["MSFT"].corr(returns["IBM"]))
```
